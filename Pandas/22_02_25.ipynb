# Pandas


```python
import pandas as pd
```

```python
data = [1,2,3,4,5,6]
s1 = pd.Series(data,)
```

```python
print(s1)
```

```python
s2 = pd.Series(data,index = ['a','b','c','d','e','f'],name = 'Numbers')
print(s2)
```

```python
data = {'a':10,'b':20,'c':30}
s3 = pd.Series(data)
print(s3)
```

```python
s4 = pd.Series([10,20,30,40,50,60],index = ['a','b','c','d','e','f'],name = 'Scores')
print(s4.index)
print(s4.values)
print(s4.name)
print(s4.dtype)
print(s4.shape)
type(s4.shape)
```

### Series Indexing and Access

```python
#LAbel based index
print(s4[0])
print(s4['a'])
```

### positional Indexing

```python
print(s4.values[3])
```

### iloc Indexing

```python
print(s4.iloc[1:4])
```

### .loc[] uses the Label/Index name, not position

```python
print(s4[['b','d']])
print(s4.loc[['b','d']])
```

```python
s4.loc[[1,3]]
```
